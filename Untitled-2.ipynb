```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.special as sp

x = np.linspace(0, 10, 10000)  

bessel_y = sp.jv(1/4, x**2 / 4)  
b = 0.10
c = 10.0
hypergeom_y = sp.hyp1f1(b, c, 1j * x**2 / 2) 
#Bessel.
plt.figure(figsize=(10, 6))
plt.plot(x, bessel_y, label=r'$J_{\,\frac{1}{4}}\left(\frac{x^2}{4}\right)$')
plt.xlabel(r'$x$')
plt.ylabel(r'$y$')
plt.title(r'Bessel Function of the First Kind $J_{\,\frac{1}{4}}\left(\frac{x^2}{4}\right)$')
plt.legend()
plt.grid(True)
plt.savefig("bessel_function.png", dpi=300) 
plt.show()
#Confluent Hypergeom.
plt.figure(figsize=(10, 6))
plt.plot(x, hypergeom_y.real, label=r'Real Part of $M(b, c; \frac{ix^2}{2})$', color='orange')  
plt.plot(x, hypergeom_y.imag, label=r'Imaginary Part of $M(b, c; \frac{ix^2}{2})$', color='blue') 
plt.xlabel(r'$x$')
plt.ylabel(r'$y$')
plt.title(r'Hypergeometric Confluent Function $M(b, c; \frac{ix^2}{2})=_1F_1(b, c; \frac{ix^2}{2})$')
plt.legend()
plt.grid(True)
plt.savefig("hypergeometric_function.png", dpi=300)  
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots()

xi_values = np.linspace(-4, 4, 8)
eta_values = np.linspace(-4, 4, 8)
y_values = np.linspace(-20, 20, 1000)


for xi in xi_values:
    x = (xi**2 - (y_values / xi)**2) / 2
    ax.plot(x, y_values, color='gray') 

for eta in eta_values:
    x = ((y_values / eta)**2 - eta**2) / 2
    ax.plot(x, y_values, color='gray')  

ax.axhline(0, color='black', linestyle='-')  
ax.axvline(0, color='black', linestyle='-')  

ax.set_aspect('equal')
ax.set_xlim([-10, 10])
ax.set_ylim([-10, 10])

ax.grid(False)

ax.spines['left'].set_position('zero')
ax.spines['left'].set_color('black')
ax.spines['left'].set_linewidth(0.5)

ax.spines['bottom'].set_position('zero')
ax.spines['bottom'].set_color('black')
ax.spines['bottom'].set_linewidth(0.5)

ax.spines['right'].set_color('none')
ax.spines['top'].set_color('none')

ax.xaxis.set_ticks_position('bottom')
ax.yaxis.set_ticks_position('left')

ax.tick_params(axis='x', colors='black', labelsize=6)
ax.tick_params(axis='y', colors='black', labelsize=6)
plt.title("Quadratic Parabolic Coordinates")

plt.show()
```
