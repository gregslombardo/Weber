```python
import math as m
import numpy as np
import scipy.special as sp
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,16)
y = sp.hyp1f1(-4.5,1,x)
plt.plot(x,y)
plt.xlim([0,x[-1]]),plt.ylim([-120,120])
plt.grid()
plt.show()
```

```python
a = np.arange(-1,1.1,0.1)
b = np.arange(0.1,0.6,0.1)
x = 1.0
print('x=',x)
for ai in a:
    print(ai,[sp.hyp1f1(ai,bi,x) for bi in b])
x = 2.0
print('x=',x)
for ai in a:
    print(ai,[sp.hyp1f1(ai,bi,x) for bi in b])
x = 10.0
print('x=',x)
for ai in a:
    print(ai,[sp.hyp1f1(ai,bi,x) for bi in b])
```
