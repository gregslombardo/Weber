```python

import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots()

xi_values = np.array([(2)**(1/2), 2*(2)**(1/2), 3*(2)**(1/2) ])
eta_values = np.array([(2)**(1/2), 2*(2)**(1/2), 3*(2)**(1/2) ])
y_values = np.linspace(-20, 20, 1000)

center_eta= y_values[650]
center_xi= (-(center_eta / xi_values[1])**2 + xi_values[1]**2) / 2

i_hat=np.array([1,0])
j_hat=np.array([0,1])

parb_sca_fact = ((center_xi)**2 + (center_eta)**2)**(1 / 2)

xi_hat = (center_xi * i_hat + center_eta * j_hat) / parb_sca_fact
eta_hat = (-center_eta * i_hat + center_xi * j_hat) / parb_sca_fact

ax.quiver(center_xi, center_eta, xi_hat[0], xi_hat[1], color='black', angles='xy', scale_units='xy', scale=0.8, width=0.004)
ax.quiver(center_xi, center_eta, eta_hat[0], eta_hat[1], color='black', angles='xy', scale_units='xy', scale=0.8, width=0.004)
ax.text(center_xi + xi_hat[0] * 0.5 + 0.55, 
        center_eta + xi_hat[1] * 0.5 + 0.3, 
        r'$\hat{\xi} \!\!\! \hat{\xi}$', 
        color='black', fontsize=8, ha='center')

ax.text(center_xi + eta_hat[0] * 0.5 - 0.65, 
        center_eta + eta_hat[1] * 0.5 - 0.6, 
        r'$\hat{\eta} \!\!\! \hat{\eta}$', 
        color='black', fontsize=8, ha='center')


for xi in xi_values:
    x = (xi**2 - (y_values / xi)**2) / 2
    ax.plot(x, y_values, color='red') 

for eta in eta_values:
    x = ((y_values / eta)**2 - eta**2) / 2
    ax.plot(x, y_values, color='green')  

circles = [plt.Circle((0, 0), radius, color='black', fill=False, linestyle=':', linewidth=1.8) for radius in [1, 4, 9]]

angular_offsets = [np.pi/4+np.pi/20, np.pi/2-np.pi/10, np.pi/2-np.pi/16]  
radial_factors = [1.2, 1.02, 1.01]  

for circle, angle, factor in zip(circles, angular_offsets, radial_factors):
    ax.add_artist(circle)
    
    x_label = circle.radius * factor * np.cos(angle)
    y_label = circle.radius * factor * np.sin(angle)

    ax.text(x_label, y_label, r'$R={}$'.format(int(circle.radius)), 
            color='black', fontsize=6, ha='left', va='bottom')

ax.text(7.5, 10,  r'$\xi=3\sqrt{2}$' , fontsize=6, ha='center', fontweight='bold')
ax.text(-3.4, 9.5,  r'$\xi=2\sqrt{2}$' , fontsize=6, ha='center', fontweight='bold')
ax.text(-10, 7.2,  r'$\xi=\sqrt{2}$' , fontsize=6, ha='center', fontweight='bold')
ax.text(-5.25, -9.5,  r'$\eta=3\sqrt{2}$' , fontsize=6, ha='center', fontweight='bold')
ax.text(3.45, -10,  r'$\eta=2\sqrt{2}$' , fontsize=6, ha='center', fontweight='bold')
ax.text(10, -7.6,  r'$\eta=\sqrt{2}$' , fontsize=6, ha='center', fontweight='bold')



ax.axhline(0, color='black', linestyle='-')  
ax.axvline(0, color='black', linestyle='-')  
ax.set_aspect('equal')
ax.set_xlim([-11, 11])
ax.set_ylim([-11, 11])
ax.grid(False)
ax.spines['left'].set_position('zero')
ax.spines['bottom'].set_position('zero')
ax.spines['right'].set_color('none')
ax.spines['top'].set_color('none')
ax.xaxis.set_ticks_position('bottom')
ax.yaxis.set_ticks_position('left')
ax.tick_params(axis='x', colors='black', labelsize=5)
ax.tick_params(axis='y', colors='black', labelsize=5)
plt.title(r'Quadratic Parabolic Coordinates')


plt.savefig('Quadratic_Parabolic.pdf', dpi=300)  
```

```python
import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots()

xi_values = np.array([(2)**(1/2), 2*(2)**(1/2), 3*(2)**(1/2) ])
xi_values_grid = np.array([ 4*(2)**(1/2),5*(2)**(1/2)])
eta_values = np.array([(2)**(1/2), 2*(2)**(1/2), 3*(2)**(1/2)  ])
eta_values_grid = np.array([4*(2)**(1/2),5*(2)**(1/2)])
y_values = np.linspace(-20, 20, 1000)

center_eta= y_values[520]
center_xi= (-(center_eta**2 / (2*xi_values[1])) + xi_values[1]) / (2)**(1/2)

i_hat=np.array([1,0])
j_hat=np.array([0,1])

parb_sca_fact = ((center_xi) + (center_eta))**(1 / 2)

xi_hat = (center_xi * i_hat + center_eta * j_hat) / parb_sca_fact
eta_hat = (-center_eta * i_hat + center_xi * j_hat) / parb_sca_fact

ax.quiver(center_xi, center_eta, xi_hat[0], xi_hat[1], color='black', angles='xy', scale_units='xy', scale=1.35, width=0.006)
ax.quiver(center_xi, center_eta, eta_hat[0], eta_hat[1], color='black', angles='xy', scale_units='xy', scale=1.35, width=0.006)
ax.text(center_xi + xi_hat[0] * 0.5 + 0.25, 
        center_eta + xi_hat[1] * 0.5 - 0.3, 
        r'$\hat{\xi} \!\!\! \hat{\xi}$', 
        color='black', fontsize=8, ha='center',
        rotation=30)  

ax.text(center_xi + eta_hat[0] * 0.5-0.2 , 
        center_eta + eta_hat[1] * 0.5+0.30 , 
        r'$\hat{\eta} \!\!\! \hat{\eta}$', 
        color='black', fontsize=8, ha='center',
        rotation=30)

for xi in xi_values:
    x = (xi - (y_values**2) / (2*xi)) / 2**(1/2)
    ax.plot(x, y_values, color='red')  

for eta in eta_values:
    x = ((y_values**2) / (2*eta) - eta) / 2**(1/2)
    ax.plot(x, y_values, color='green')  

for xi in xi_values_grid:
    x = (xi - (y_values**2) / (2*xi)) / 2**(1/2)
    ax.plot(x, y_values, color='gray')  

for eta in eta_values_grid:
    x = ((y_values**2) / (2*eta) - eta) / 2**(1/2)
    ax.plot(x, y_values, color='gray')      

ax.axhline(0, color='black', linestyle='-')  
ax.axvline(0, color='black', linestyle='-')  

ax.set_aspect('equal')
ax.set_xlim([-4, 4])
ax.set_ylim([-4, 4])

ax.grid(False)
circles = [plt.Circle((0, 0), radius, color='black', fill=False, linestyle=':', linewidth=1.4) for radius in [1, 2, 3]]

angular_offsets = [np.pi/2-np.pi/32, np.pi/3+np.pi/16, np.pi/3+np.pi/50]  
radial_factors = [1.05, 1.005, 1.005]  

for circle, angle, factor in zip(circles, angular_offsets, radial_factors):
    ax.add_artist(circle)
    
    x_label = circle.radius * factor * np.cos(angle)
    y_label = circle.radius * factor * np.sin(angle)

    ax.text(x_label, y_label, r'$\rho={}$'.format(int(circle.radius)), 
            color='black', fontsize=6, ha='left', va='bottom')

ax.text(2.59,3.1,  r'$\xi=3\sqrt{2}$' , fontsize=5.5, ha='center', fontweight='bold')
ax.text(1, 3.35,  r'$\xi=2\sqrt{2}$' , fontsize=5.5, ha='center', fontweight='bold')
ax.text(-1.3, 3.3,  r'$\xi=\sqrt{2}$' , fontsize=5.5, ha='center', fontweight='bold')
ax.text(-2.72, -3.2,  r'$\eta=3\sqrt{2}$' , fontsize=6, ha='center', fontweight='bold')
ax.text(-0.99, -3.7,  r'$\eta=2\sqrt{2}$' , fontsize=6, ha='center', fontweight='bold')
ax.text(1.3, -3.45,  r'$\eta=\sqrt{2}$' , fontsize=6, ha='center', fontweight='bold')    

ax.spines['left'].set_position('zero')
ax.spines['left'].set_color('black')
ax.spines['left'].set_linewidth(0.5)

ax.spines['bottom'].set_position('zero')
ax.spines['bottom'].set_color('black')
ax.spines['bottom'].set_linewidth(0.5)

ax.spines['right'].set_color('none')
ax.spines['top'].set_color('none')


current_labels = [item.get_text() for item in ax.get_xticklabels()]
ax.set_xticklabels(current_labels, ha='left')  

current_y_labels = [item.get_text() for item in ax.get_yticklabels()]
ax.set_yticklabels(current_y_labels, va='bottom')  


ax.xaxis.set_ticks_position('bottom')
ax.yaxis.set_ticks_position('left')

ax.tick_params(axis='x', colors='black', labelsize=7)
ax.tick_params(axis='y', colors='black', labelsize=7)
plt.title("Linear Parabolic Coordinates")


plt.savefig('Linear_Parabolic.pdf', dpi=300)  

plt.show()
```
